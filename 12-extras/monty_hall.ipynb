```python
import random
from tqdm import tqdm
```

```python
portas = [1,2,3]
```

```python
def sorteio_porta(portas):
  return random.choice(portas)
```

```python
ganhou_original=0
perdeu_original=0
ganhou_troca=0
perdeu_troca=0

for i in tqdm(range(1000000)):
  porta_premio=sorteio_porta(portas)
  porta_jogador=sorteio_porta(portas)

  opcoes=portas.copy()
  if porta_jogador == porta_premio:
    opcoes.remove(porta_jogador)
  else:
    opcoes.remove(porta_premio)
    opcoes.remove(porta_jogador)

  porta_apresentador = sorteio_porta(opcoes)

  segunda_fase=portas.copy()
  segunda_fase.remove(porta_apresentador)

  if porta_jogador == porta_premio:
    ganhou_original+=1
  else:
    perdeu_original+=1

  segunda_fase.remove(porta_jogador)
  if segunda_fase[0]==porta_premio:
    ganhou_troca+=1
  else:
    perdeu_troca+=1
print()
print(ganhou_original/(ganhou_original+perdeu_original))
print(ganhou_troca/(ganhou_troca+perdeu_troca))
```
